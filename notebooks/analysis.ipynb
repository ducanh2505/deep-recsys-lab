# Multi-VAE run analysis

This notebook only reads outputs produced by `deep-recsys-lab`; it does not download data or train a model. Set `RUN_DIR` to a run directory containing `metrics.jsonl` and, optionally, `test_metrics.json`.

```python
import json
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd

RUN_DIR = Path("outputs/multvae")
metrics_path = RUN_DIR / "metrics.jsonl"
if not metrics_path.exists():
    raise FileNotFoundError(f"Run metrics not found: {metrics_path}")
history = pd.DataFrame(json.loads(line) for line in metrics_path.read_text().splitlines())
history.tail()
```

```python
metric_columns = [column for column in history.columns if column.startswith("validation/")]
ax = history.plot(x="epoch", y=metric_columns, marker="o", figsize=(10, 5))
ax.set_title("Validation ranking metrics")
ax.set_ylabel("score")
ax.grid(True, alpha=0.25)
plt.show()
```

```python
test_path = RUN_DIR / "test_metrics.json"
if test_path.exists():
    display(pd.Series(json.loads(test_path.read_text()), name="test"))
else:
    print("No test_metrics.json was saved for this run.")
```
